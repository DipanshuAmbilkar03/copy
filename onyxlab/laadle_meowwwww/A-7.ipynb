```python
"""A-7: ANN training with Forward Propagation and Back Propagation (from scratch)."""

import numpy as np

# 1. Dataset (XOR problem)
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([[0], [1], [1], [0]])

# 2. Initialize weights and biases
np.random.seed(42)
input_neurons = 2
hidden_neurons = 4
output_neurons = 1

W1 = np.random.uniform(size=(input_neurons, hidden_neurons))
b1 = np.random.uniform(size=(1, hidden_neurons))
W2 = np.random.uniform(size=(hidden_neurons, output_neurons))
b2 = np.random.uniform(size=(1, output_neurons))

# 3. Activation function and its derivative
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# 4. Training loop (Forward & Backward Propagation)
epochs = 10000
lr = 0.1

print("Training ANN from scratch...")
for epoch in range(epochs):
    # --- Forward Propagation ---
    hidden_layer_activation = np.dot(X, W1) + b1
    hidden_layer_output = sigmoid(hidden_layer_activation)
    
    output_layer_activation = np.dot(hidden_layer_output, W2) + b2
    predicted_output = sigmoid(output_layer_activation)
    
    # --- Backpropagation ---
    error = y - predicted_output
    d_predicted_output = error * sigmoid_derivative(predicted_output)
    
    error_hidden_layer = d_predicted_output.dot(W2.T)
    d_hidden_layer = error_hidden_layer * sigmoid_derivative(hidden_layer_output)
    
    # --- Update Weights and Biases ---
    W2 += hidden_layer_output.T.dot(d_predicted_output) * lr
    b2 += np.sum(d_predicted_output, axis=0, keepdims=True) * lr
    W1 += X.T.dot(d_hidden_layer) * lr
    b1 += np.sum(d_hidden_layer, axis=0, keepdims=True) * lr

    if epoch % 2000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}: Loss = {loss:.4f}")

# 5. Final Output
print("\nFinal Predictions for XOR Dataset:")
# print((np.round(predicted_output)))

for i in range(len(X)):
    print(X[i], "->", round(predicted_output[i][0]))
```
