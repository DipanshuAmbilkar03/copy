```python
import numpy as np
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])
y= np.array([[0],[1],[1],[0]]) 

np.random.seed(42) 
i = 2 
h = 4 
o = 1 

W1 = np.random.uniform(size=(i,h))
b1 = np.random.uniform(size=(1,h))
W2 = np.random.uniform(size=(h,o))
b2 = np.random.uniform(size=(1,o)) 

def sigmoid(n) : 
    return 1 / (1 + np.exp(-n))

def sigmoid_der(x) : 
    return x * (1 - x) 

epoch = 10000 
lr = 0.1 

for e in range(epoch) : 
    hiddenlayer = np.dot(X,W1) + b1
    hiddenOutput  = sigmoid(hiddenlayer)

    outputLayer = np.dot(hiddenOutput,W2) + b2 
    pred = sigmoid(outputLayer)

    err = y - pred 

    # backpropagation
    d_pred = err * sigmoid_der(pred)
    error_hidden  = d_pred.dot(W2.T)
    d_hidden = error_hidden * sigmoid_der(hiddenOutput)

    W2+=hiddenOutput.T.dot(d_pred)*lr 
    b2+=np.sum(d_pred,axis=0,keepdims=True) *lr 

    W1+=X.T.dot(d_hidden) * lr 
    b1+=np.sum(d_hidden,axis=0,keepdims=True)*lr
    
print(np.round(pred))
```
