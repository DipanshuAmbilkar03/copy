```python
import numpy as np
def ascii(ch) :
    return [int(bit) for bit in format(ord(ch),'08b')]

digit = [str(i) for i in range(10)]

X = np.array([ascii(d) for d in digit])
Y = np.array([int(d) % 2 for d in digit])

print(X)
print(digit)
print(Y)

w = np.zeros(8)
lr = 0.1 
b = 0
```

```python
for epoch in range(100) : 
    error = 0
    for x,y in zip(X,Y) :
        net = np.dot(w,x) + b 
        pred = 1 if net >= 0 else 0 

        update = (y - pred) * lr
        w += update *x 
        b += update 
        error += abs(update)
    if(error == 0) : 
        print("Traing complete.",epoch+1,"epoch")
```

```python
correct = 0
for d,x,y in zip(digit,X,Y) : 
    net = np.dot(w,x) + b 
    pred = 1 if net >= 0 else 0 
    
    res = "odd" if pred == 1 else "even" 

    print(d," ->",ord(d)," ->", res) 

    if(pred == y) : 
        correct += 1 

acc = correct / len(digit) 
print(int(acc*100))
```
