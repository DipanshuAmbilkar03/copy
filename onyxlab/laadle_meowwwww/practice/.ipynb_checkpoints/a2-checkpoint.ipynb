```python
import numpy as np
def andnot(x,weight): 
    s = np.dot(x,weight)

    if(s >= 0 ) :
        return 1
    else :
        return 0
def and_(x,weight): 
    s = np.dot(x,weight)

    if(s >= 2 ) :
        return 1
    else :
        return 0 
        
def or_(x,weight): 
    s = np.dot(x,weight)

    if(s >= 0 ) :
        return 1
    else :
        return 0

val = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

weight = [-1,1]
for i in val : 
    print(i[0],i[1],"->",andnot(i,weight))
print("-"*50)

weight = [1,1]
for i in val : 
    print(i[0],i[1],"->",and_(i,weight))
print("-"*50) 

weight = [1,1]
for i in val : 
    print(i[0],i[1],"->",or_(i,weight))
print("-"*50)
```
