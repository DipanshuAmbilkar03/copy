```python
import numpy as np
import matplotlib.pyplot as plt
def sigmoid(n) :
    return 1/(1+np.exp(-n))
def Bin_linear(n) :
    if n >= 0 :return 1 
    elif n < 0 : return 0
def tanh(n) :
    return np.tanh(n)
def relu(n) :
    return np.maximum(0,n)
def leaky_relu(n) :
    return np.maximum(0.5*n,n)
def softmax(n) : 
    return np.log1p(np.exp(n))
x = np.linspace(-8,8,5000)
# print(x)
# plt.plot(x)
```

```python
plt.figure(figsize=(6,4))
plt.plot(x,sigmoid(x),label='sigmoid')
```

```python
plt.figure(figsize=(6,4))
plt.plot(x,tanh(x),label='sigmoid')
```

```python
plt.figure(figsize=(6,4))
plt.plot(x,relu(x),label='sigmoid')
```

```python
plt.figure(figsize=(6,4))
plt.plot(x,leaky_relu(x),label='sigmoid')
```

```python
plt.figure(figsize=(6,4))
plt.plot(x,softmax(x),label='sigmoid')
```
