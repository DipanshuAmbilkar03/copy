# Experiment A-4


```python
import numpy as np
import matplotlib.pyplot as plt

# Training Data
X = np.array([
    [1, 1],
    [2, 1],
    [1, 2],
    [2, 2],
    [4, 4],
    [5, 4],
    [4, 5],
    [5, 5]
])

Y = np.array([0, 0, 0, 0, 1, 1, 1, 1])

# Initial weights and bias
w = np.zeros(2)
b = 0
lr = 0.1

# Training
for epoch in range(20):

    errors = 0

    for x, y in zip(X, Y):

        net = np.dot(w, x) + b

        if net >= 0:
            y_pred = 1
        else:
            y_pred = 0

        update = lr * (y - y_pred)

        w = w + update * x
        b = b + update

        if update != 0:
            errors += 1

    print("Epoch", epoch + 1, "Errors =", errors)

    if errors == 0:
        break

# Create decision region
x_min, x_max = 0, 6
y_min, y_max = 0, 6

xx, yy = np.meshgrid(
    np.linspace(x_min, x_max, 200),
    np.linspace(y_min, y_max, 200)
)

grid = np.c_[xx.ravel(), yy.ravel()]

Z = []

for point in grid:

    net = np.dot(w, point) + b

    if net >= 0:
        Z.append(1)
    else:
        Z.append(0)

Z = np.array(Z).reshape(xx.shape)

# Plot decision regions
plt.contourf(xx, yy, Z, alpha=0.3)

# Plot points
plt.scatter(X[Y == 0][:, 0], X[Y == 0][:, 1], label="Class 0")
plt.scatter(X[Y == 1][:, 0], X[Y == 1][:, 1], label="Class 1")

# Decision boundary line
x1 = np.linspace(0, 6, 100)
x2 = -(w[0] * x1 + b) / w[1]

plt.plot(x1, x2, 'k--', label="Decision Boundary")

plt.xlabel("X1")
plt.ylabel("X2")
plt.title("Perceptron Decision Region")

plt.legend()
plt.grid(True)

plt.show()
```
