# Experiment 7: Text Analytics

**Aim:**
1. Extract a sample document and apply document preprocessing: Tokenization, POS Tagging, Stop Words Removal, Stemming, and Lemmatization.
2. Create a representation of documents by calculating **Term Frequency (TF)** and **Inverse Document Frequency (IDF)**.

### Step 1: Install and Import Libraries

```python
import nltk
import math
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download required NLTK data
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')

print('Libraries imported successfully!')
```

### Step 2: Sample Document
We define a sample text document that we will preprocess.

```python
document = """
Natural Language Processing is a fascinating field of Artificial Intelligence.
It helps computers understand, interpret, and generate human language.
Text mining and machine learning are used together to extract useful information.
The running dogs are barking loudly near the tall buildings.
"""

print('Sample Document:')
print(document)
```

### Step 3: Tokenization
**Tokenization** splits the text into individual words (tokens). This is the first step in text preprocessing.

```python
tokens = word_tokenize(document)

print('Tokens:')
print(tokens)
print(f'\nTotal tokens: {len(tokens)}')
```

### Step 4: POS Tagging
**Part-of-Speech (POS) Tagging** assigns a grammatical label to each word (e.g., Noun, Verb, Adjective).

Common POS tags: `NN` = Noun, `VB` = Verb, `JJ` = Adjective, `RB` = Adverb.

```python
pos_tags = pos_tag(tokens)

print('POS Tags (word, tag):')
for word, tag in pos_tags:
    print(f'  {word:20s} -> {tag}')
```

### Step 5: Stop Words Removal
**Stop words** are common words like 'the', 'is', 'and' that carry little meaning. We remove them to focus on meaningful words.

```python
stop_words = set(stopwords.words('english'))

# Keep only alphabetic tokens that are NOT stop words
filtered_tokens = [word for word in tokens if word.isalpha() and word.lower() not in stop_words]

print('Tokens after Stop Words Removal:')
print(filtered_tokens)
print(f'\nTotal after filtering: {len(filtered_tokens)}')
```

### Step 6: Stemming
**Stemming** reduces a word to its root/base form by removing suffixes.
Example: `running` → `run`, `barking` → `bark`

We use the **Porter Stemmer**, which is the most common stemming algorithm.

```python
stemmer = PorterStemmer()

stemmed_tokens = [stemmer.stem(word) for word in filtered_tokens]

print('Original  ->  Stemmed:')
for original, stemmed in zip(filtered_tokens, stemmed_tokens):
    print(f'  {original:20s} -> {stemmed}')
```

### Step 7: Lemmatization
**Lemmatization** also reduces words to their base form, but it uses vocabulary and grammar rules to return the proper dictionary form.
Example: `running` → `run`, `better` → `good`

Lemmatization is more accurate than Stemming because it considers context.

```python
lemmatizer = WordNetLemmatizer()

lemmatized_tokens = [lemmatizer.lemmatize(word.lower()) for word in filtered_tokens]

print('Original  ->  Lemmatized:')
for original, lemmatized in zip(filtered_tokens, lemmatized_tokens):
    print(f'  {original:20s} -> {lemmatized}')
```

## Part 2: TF-IDF (Term Frequency – Inverse Document Frequency)

TF-IDF is used to represent how important a word is in a document relative to a collection of documents (corpus).

- **TF (Term Frequency)** = (Number of times word appears in doc) / (Total words in doc)
- **IDF (Inverse Document Frequency)** = log(Total docs / Number of docs containing the word)
- **TF-IDF** = TF × IDF

### Step 8: Define the Corpus (Collection of Documents)

```python
corpus = [
    "Natural Language Processing is a field of Artificial Intelligence",
    "Machine learning and text mining extract useful information from data",
    "Deep learning is used in Natural Language Processing and computer vision",
    "Text mining uses machine learning algorithms to find patterns in text"
]

print('Corpus (4 documents):')
for i, doc in enumerate(corpus, 1):
    print(f'  Doc {i}: {doc}')
```

### Step 9: Compute Term Frequency (TF)

```python
def compute_tf(doc):
    words = doc.lower().split()
    tf = {}
    for word in words:
        tf[word] = tf.get(word, 0) + 1
    # Divide by total words to get frequency
    total = len(words)
    for word in tf:
        tf[word] = tf[word] / total
    return tf

# Compute TF for each document
tf_list = [compute_tf(doc) for doc in corpus]

print('Term Frequency for Document 1:')
for word, score in sorted(tf_list[0].items()):
    print(f'  {word:25s}: {score:.4f}')
```

### Step 10: Compute Inverse Document Frequency (IDF)

```python
def compute_idf(corpus):
    total_docs = len(corpus)
    idf = {}
    
    # Find all unique words
    all_words = set(word.lower() for doc in corpus for word in doc.split())
    
    for word in all_words:
        # Count how many documents contain this word
        docs_with_word = sum(1 for doc in corpus if word in doc.lower().split())
        idf[word] = math.log(total_docs / docs_with_word)
    return idf

idf = compute_idf(corpus)

print('IDF values for all unique words:')
for word, score in sorted(idf.items()):
    print(f'  {word:25s}: {score:.4f}')
```

### Step 11: Compute TF-IDF Score

```python
def compute_tfidf(tf, idf):
    tfidf = {}
    for word, tf_score in tf.items():
        tfidf[word] = tf_score * idf.get(word, 0)
    return tfidf

# Compute TF-IDF for each document
tfidf_list = [compute_tfidf(tf, idf) for tf in tf_list]

# Print TF-IDF scores for Document 1
print('TF-IDF Scores for Document 1:')
for word, score in sorted(tfidf_list[0].items(), key=lambda x: -x[1]):
    print(f'  {word:25s}: {score:.4f}')
```

### Step 12: TF-IDF for All Documents (Summary Table)

```python
import pandas as pd

# Build a matrix: rows = documents, columns = words
all_words_sorted = sorted(idf.keys())
rows = []

for i, tfidf in enumerate(tfidf_list):
    row = {'Document': f'Doc {i+1}'}
    for word in all_words_sorted:
        row[word] = round(tfidf.get(word, 0.0), 4)
    rows.append(row)

tfidf_df = pd.DataFrame(rows).set_index('Document')

print('TF-IDF Matrix (Documents x Words):')
display(tfidf_df)
```
