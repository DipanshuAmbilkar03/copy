# Experiment 3: Measures of Central Tendency and Variability
This notebook computes summary statistics grouped by a categorical variable and basic statistical details for different species in the Iris dataset.

```python
import pandas as pd
import numpy as np
```

### Load the Dataset
We load `Iris.csv` into a Pandas DataFrame.

```python
# Load the dataset
df = pd.read_csv('Iris.csv')

# Display the first 5 rows
df.head()
```

### Part 1: Summary Statistics Grouped by Categorical Variable
We group the numeric variables by the categorical variable `Class Label` (which represents the species) and compute summary statistics including mean, median, min, max, and standard deviation.

```python
# Drop the 'Id' column as it's not a numeric measurement
df_numeric = df.drop(columns=['Id'])

# Group by the categorical variable 'Class Label'
grouped = df_numeric.groupby('Class Label')

# Calculate summary statistics: mean, median, min, max, std
summary_stats = grouped.agg(['mean', 'median', 'min', 'max', 'std'])

print("Summary Statistics Grouped by 'Class Label':")
display(summary_stats)
```

#### Create a numeric list for each categorical response
We map each unique category in `Class Label` to a numeric value and create a list containing these mapped values for each row in the dataset.

```python
# Identify unique categories
categories = df['Class Label'].unique()

# Create a dictionary mapping each category to a numeric value (0, 1, 2)
category_mapping = {category: idx for idx, category in enumerate(categories)}
print("Mapping of categories to numeric values:", category_mapping)

# Create a list containing the numeric value for each response to the categorical variable
numeric_response_list = [category_mapping[val] for val in df['Class Label']]

# Display the total length and the first 15 elements of the list
print(f"\nLength of numeric list: {len(numeric_response_list)}")
print("First 15 numeric values in the list:")
print(numeric_response_list[:15])
```

### Part 2: Basic Statistical Details of Each Species
We filter the dataset for `Iris-setosa`, `Iris-versicolor`, and `Iris-virginica` and use the `describe()` function to display percentiles, mean, standard deviation, and more.

```python
# The unique species present in the dataset
species_list = df['Class Label'].unique()

for species in species_list:
    print(f"\n{'='*40}")
    print(f"Statistical Details for {species}")
    print(f"{'='*40}")
    
    # Filter dataframe for the current species
    species_df = df_numeric[df_numeric['Class Label'] == species]
    
    # describe() computes count, mean, std, min, 25%, 50%, 75%, and max
    display(species_df.describe())
    print("\n")
```
