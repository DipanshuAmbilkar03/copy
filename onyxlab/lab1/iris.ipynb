### 5. Data Formatting and Normalization
We summarize the data types using `df.dtypes`. If a variable is stored as the wrong type, we convert it using `astype()`. Here, we demonstrate type conversion by changing the first float column to `float32`.

```python

print('Initial statistics:')
display(df.describe())
```

```python
# Check data types of all variables
print('Data types of variables:')
print(df.dtypes)
```

```python

# Example of type conversion (converting float64 to float32 to save memory)
float_cols = df.select_dtypes(include=['float64']).columns
if len(float_cols) > 0:
    col_to_convert = float_cols[0]
    df[col_to_convert] = df[col_to_convert].astype('float32')
    print(f'Data types after converting {col_to_convert} to float32:')
    print(df.dtypes)
else:
    print('No float64 columns found to convert.')
```

### 6. Turn Categorical Variables into Quantitative Variables
Machine learning models often require numerical inputs. We use `LabelEncoder` to convert the categorical variable (`Species`) into a quantitative/numerical format.

```python
# Create a label encoder object
le = LabelEncoder()

cat_cols = df.select_dtypes(include=['object']).columns
if len(cat_cols) > 0:
    cat_col = cat_cols[0]
    # Fit and transform the categorical column
    df[f'{cat_col}_encoded'] = le.fit_transform(df[cat_col])
    
    print(f'Mapping of {cat_col} to integers:')
    for idx, class_name in enumerate(le.classes_):
        print(f'{class_name} -> {idx}')
        
    print('-'*30)
    print('Dataframe with new quantitative variable:')
    display(df.head())
else:
    print('No categorical columns found.')
```
